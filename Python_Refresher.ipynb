```python
import pandas as pd

gobachop_data = pd.DataFrame({
    "Vendor": ["Ma Musu", "Sia", "Flomo"],
    "Sales": [4000, 9500, 1500]
})

# Example of attributes: .shape tells us the of the DataFrame
print("DataFrame shape:", gobachop_data.shape)

# Example of a method: .head() shows the first few rows
print(gobachop_data.head)

# Example of the method: .describe() gives statistics for numeric columns
print(gobachop_data.describe())
```
